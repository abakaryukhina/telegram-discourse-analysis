```python
#Schritt 2. Strukturierung der Daten mit pandas.

# Importierung der notwendigen Bibliotheken
import pandas as pd
import json
from pprint import pprint
import emoji              #Damit ich die zahlreichen Emojis entfernen kann

filepath = "../data/Studienleistung/cs_association_0.json" #Nach dem Data-Export habe ich die Datei in den Ordner python_corpuslinguistic/data/Studienleistung gelegt.
with open (filepath, "r", encoding = "utf-8") as file:
    dictionary_list = json.load(file) #Da telegram-download-chat default-weise eine json- und txt-Datei als output liefert, musste ich die json-Datei mithilfe der Python-Methode json.load öffnen.
    
usernames = []
dates = []
messages = []
translations = []
for dictionary in dictionary_list: #Die Struktur der Output-Datei ist eine Liste von Dictionaries (siehe cs_association_0.json)
    if dictionary.get("message", ""): #Da anscheinend nicht jede Listeneinheit (dictionary) einen Schlüssel "message" hat, 
                                        #stürzt das Programm ab, wenn man versucht, ohne eine weitere Bedingung den Wert des Schlüssels "message" der Liste "messages" hinzuzufügen.
                                        #Darüber hinaus kann der Wert leer sein, wenn es doch einen "message"-Schlüssel gibt. 
                                        #Mit der Bedingung "if dictionary.get("message", ""):" stelle ich sicher, dass der Code nur die Listen durchläuft, die den Schlüssel "message" hat, 
                                        #und sein Wert nicht leer ist.
        usernames.append(dictionary["user_display_name"]) 
        dates.append(dictionary["date"])
        message = dictionary["message"].replace("\n", " ")
        message = emoji.replace_emoji(message, replace="") #Entfernung der Emojis
        messages.append(message)
        translations.append("") #Ich erstelle eine zusätzliche Spalte, die ich für Übersetzungen ins Deutsche freihalte.

data = {"username": usernames,
        "dates": dates,
        "messages": messages,
       "translations": translations}
df = pd.DataFrame(data) #Erstellung einer Pandas-Tabelle
display(df)
df.to_csv("test_df_cs_association_0.csv", sep = "\t") #Speicherung der Pandas-Tabelle unter "test_df_cs_association_0.csv"
```

```python
#Schritt 3. Übersetzung der Messages via GoogleTranslator
from deep_translator import GoogleTranslator    #Übersetzung der Nachrichten erfollgt mittels der Bibliothek deep_translator 
                                                #(https://github.com/prataffel/deep_translator#deepltranslator).
                                                #Der ausgewählte Übersetzter ist GoogleTranslator, weil man keine zusätzliche API anfragen muss und
                                                #keine limitierte Anzahl der zu übersetzenden Zeichen hat (wie z.B. DeepL).
import time  #Notwendig, um den Übersetzer nicht zu "überlasten" (siehe unten)
import spacy 
translations_de = []
translator = GoogleTranslator(source='auto', target='de') #Die Nutzunganweisungen von deep_translator (GoogleTranslator) habe ich auf der github-Seite gefunden (oben verlinkt)
nlp = spacy.load('ru_core_news_lg')
for i, message in enumerate(messages):
    try:                                            #Die Sturzabsicherung ist notwendig, weil GoogleTranslator nur 5.000 Zeichen je Übersetzung übersetzen kann. 
                                                    #Wenn die Nachricht mehr als 5.000 Zeichen hat, stürzt das Programm ohne try-except-Methode ab. 
        result = translator.translate(text=message) #Die Nutzunganweisungen von deep_translator (GoogleTranslator) habe ich auf der github-Seite gefunden (oben verlinkt)
        translations_de.append(result) 
        time.sleep(0.5) #Nach jeder Übersetzung gebe ich den GoogleTranslor Zeit (5 Milisekunden), sich zu erholen, damit er nicht durch zu viele Anfragen abstürzt.
        
        if (i + 1) % 100 == 0: 
            print(f"✅ {i + 1} out of {len(messages)} successfully translated...")  #Da ich insgesamt fast 6.000 Nachrichten habe, baue ich diese print-Zeile ein, 
                                                                                    #damit ich in 100-Schritt sehe, wie viele Nachrichten übersetzt werden. 
                                                                                    #Bei jeder hundertsten Nachricht (Divisionsrest (%) ist 0) 
    except Exception as e:
        print(f"❌No Translation Possible For Message {i}") #Die Nachrichten, die (aufgrund der den Limit überschreitenden Zeichenanzahl) nicht übersetzt werden, 
                                                            #werden mit dieser print-Zeile entsprechend markiert
        print(f"⚠️Trying translating Message {i} one sentence at a time")
        try:
            if len(message) > 2000: #Der Limit von 5.000 Zeichen gilt für das lateinische Alphabet. 
                                    #Die kyrillischen Buchstaben, so wie ich im Internet gelesen habe, wiegen 2-3 Bytes.
                                    #Deswegen habe ich eine Sicherheitskontrolle eingebaut, die die Nachrichten mit über 2.000 Zeichen mittels spacy in Sätze zerlegen lässt 
                                    #und sie satzweise übersetzt.
                doc = nlp(message)
                sentences_list = []
                for i_sent, sentence in enumerate(doc.sents):
                    try:
                        sentence_translation = translator.translate(text=sentence.text)
                        if sentence_translation:
                            sentences_list.append(sentence_translation)
                            print(f"✅ Sentence {i_sent} of Message {i} successfully translated via GoogleTranslator")  #print-Zeile, damit ich auf Nummer sicher gehen kann, 
                                                                                                                        #dass jeder Satz übersetzt wird. Die Sicherheit wird auch 
                                                                                                                        #durch die try-except-Methode gewahrt
                        time.sleep(0.5) # Auch um auf Nummer sicher zu gehen, dass der GoogleTransaltor nicht überstrapaziert wird.
                        
                    except Exception as e:
                        print(f"❌ Sentence {i_sent} of Message {i} cannot be translated via GoogleTranslator")
                if sentences_list:       
                    translation = " ".join(sentences_list)
                    translations_de.append(translation)
                    print(f"✅ Message {i} successfully translated via GoogleTranslator")
            else:
                try:
                    result = translator.translate(text=message)
                    translations_de.append(result) 
                    time.sleep(0.5)
                    
                    if (i + 1) % 100 == 0:
                        print(f"✅ {i + 1} out of {len(messages)} successfully translated...")
                except Exception as e:
                    print(f"No Translation Possible For Message {i}")
                    translations_de.append(f"No Translation Possible For Message {i}")  #Falls keine Übersetzung möglich gewesen wäre, 
                                                                                        #wäre die dem Nachrichtenindex entsprechende Übersetzungsindex trotzdem nicht verloren gegegangen
                                                                                        #In diesem Fall hätte ich versucht, die Nachricht durch einen anderen Übersetzer übersetzten zu lassen
                                                                                        #Und den Platzhalter-Text über den entsprechenden Index durch eine neue Übersetzung zu ersetzen.
        except Exception as e:
            print(f"❌No Translation Possible For Message {i}")


print(translations_de[0]) 
```

```python
# Schritt 5. Zusammenführung der Daten, die in demselben Monat erschienen sind
df_sorted = df_with_transl.sort_values(by="dates", ascending=True) #Sortierung nach dem Erscheinungsdatum
df_sorted["dates"] = df_sorted["dates"].str[:7] #Sortierung nach dem Jahr und Monat. Die ersten sieben Zeichen des Datum-Strings beinhalten genau das Jahr und den Monat.

pub_year_month = []
pub_translated_lists = []

for date in df_sorted["dates"]:
    if date not in pub_year_month: #Damit das Datum nur einmal in der Pandas-Tabelle erscheint.
        pub_year_month.append(date)
        pub_translated_frame = df_sorted[df_sorted["dates"] == date]    #einmalige Herausfilterung aller Zeilen mit den Nachrichten, 
                                                                        #die am in der for-Schleife an die Reihe gekommenen Datum erschienen sind.
        pub_translated_list = pub_translated_frame["translations"].tolist() #Überführung der Nachrichtenspalte der herausgefilterten Zeilen in eine Liste
        pub_translated_joined = " ".join(pub_translated_list)
        pub_translated_lists.append(pub_translated_joined) 

data_sorted = {"pub_year_month": pub_year_month, # Erstellung einer neuen Pandas-Tabelle
    "pub_translated": pub_translated_lists} 

data_sorted_frame = pd.DataFrame(data_sorted)
data_sorted_frame.to_csv("data_sorted_frame_cs_association.csv", sep = "\t")
display(data_sorted_frame)
```

```python
# Schritt 6. Untersuchung der relativen und absoluten Häufigkeit der Lemmata nach dem Jahr und Monat:

from collections import Counter
nlp_de = spacy.load("de_core_news_lg")
with open ("frequency_year_month_cs_association_0.txt", "w", encoding = "utf-8") as file: #Speicherung der Daten in "frequency_year_month_cs_association_0.txt"
    for pub, year_month in zip(pub_translated_lists, pub_year_month): #Damit ich einer Frequenz-Tabelle immer das entsprechende Jahr zuordnen könnte, 
                                                                    #lasse ich das Programm parallel zwei Listen durchlaufen.
        doc_de = nlp_de(pub)
        N = len([token.text for token in doc_de if token.text])
        lemma_freq = Counter([token.lemma_ for token in doc_de if not token.is_stop and not token.is_punct and not token.is_space]).most_common(50) #Manche Einheiten, die von spacy als Tokens erkannt wurden, waren leer;
                                                                                                                                                    #Ich gehe davon aus, dass es URL-Links waren. Was sie auch immer sind, 
                                                                                                                                                    #wollte ich sie nicht im Korpus haben. Deswegen habe ich sie durch "token.is_space" ausgefiltert. 
        print (year_month, file = file)
        for lemma, freq in lemma_freq:
            ipm = (freq / N) * 1_000_000
            rel_freq = (freq / N) * 100
            print(f"{lemma:>25} || {freq:>5} ({rel_freq:.2f}%) || {ipm:>20}", file = file)
        print("\n","\n","\n", file = file)
    
```

```python
# Schritt 7. Erstellung der N-Gramme:
from pyvis.network import Network
from nltk.util import ngrams

for pub, year_month in zip(pub_translated_lists, pub_year_month):
    doc_de = nlp_de(pub)
    tokens = [token.lemma_ for token in doc_de if not token.is_stop and not token.is_punct and not token.is_space]
    bigrams = list(ngrams(tokens, 2)) # Erstellung der Bigramme
    
    #Visualisierung der Bigramme:
    
    list_strings = []
    for t in bigrams:
        string_chain = " ".join(t)
        list_strings.append(string_chain)
    top_50 = Counter(list_strings).most_common(50)
    
    net = Network()
    unique_tokens = []

    for bigram_string, freq in top_50:
        words = bigram_string.split(" ")
        if words[0] not in unique_tokens: #Sicherstellung, dass dieselben Bigramme-Elemente im "net" nicht doppelt erstellt werden.
            net.add_node(words[0])
            unique_tokens.append(words[0])
        if words[1] not in unique_tokens: #Sicherstellung, dass dieselben Bigramme-Elemente im "net" nicht doppelt erstellt werden.
            net.add_node(words[1])
            unique_tokens.append(words[1])
        net.add_edge(words[0], words[1]) # Verknüpfung beider Elemente
    net.show_buttons(filter_=['physics'])
    net.save_graph(f"my_network_cs_association_0_pubdate_{year_month}.html")
    
```
